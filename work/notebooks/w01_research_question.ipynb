## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/ak8x6/flyrank-seo-ml-pipeline"
REPO_DIR = "flyrank-seo-ml-pipeline"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir(os.path.join("..", ".."))

import pandas as pd
import numpy as np

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(f"Dataset: {df.shape[0]:,} rows x {df.shape[1]} columns")
print(f"Clients: {df['client_id'].nunique()}")

# --- NUMBER 1: Decline rate ---
decline_rate = (df["trend_direction"].str.lower() == "down").mean()
print(f"\nNumber 1 — Decline rate: {decline_rate:.1%} of pages are declining")
print(f"  {int(decline_rate * len(df)):,} out of {len(df):,} pages trending down")
print(f"  Random guessing would get it right {decline_rate:.0%} of the time")

# --- NUMBER 2: Model vs hand rule gap ---
print(f"\nNumber 2 — Starter pipeline Precision@50:")
print(f"  Hand-written rule:  0.240  (~12 of top 50 correct)")
print(f"  Random forest:      0.740  (~37 of top 50 correct)")
print(f"  Lift: ~3.1x — the model finds 3x more real declining pages")

# --- NUMBER 3: Actionable page volume ---
visible = df[df["impressions_90d"] >= 100]
stale_visible = visible[visible["days_since_last_update"] >= 180]
print(f"\nNumber 3 — Actionable pages:")
print(f"  Visible (>=100 impressions): {len(visible):,} ({len(visible)/len(df):.0%})")
print(f"  Stale + visible (180+ days old): {len(stale_visible):,}")
print(f"  These are the core pool where refresh recommendations matter most")
```
